# Trinity Admission Data Preparation <br>


The goal of this data preparation project is to ready the data for constructing a classification model to determine whether an accepted applicant will decide to attend the University, based on information collected regarding the University’s recently accepted applicants, ranging in entry term from Fall 2017, Fall 2018, Fall 2019, Fall 2020, and Fall 2021.  Your final data set should be ready for modeling. 

This script demonstrates how to clean some typical variables for the train dataset and the cleaning requiremens for all variables. 

The train dataset is a subset of the original dataset, which is used to train the model to understand the relationships between variables. Then the trained model will predict the target variable using predictors in the test dataset.

You need to handle all the columns/variables that are not processed in this script following the instructions in the comments.


```python
import pandas as pd
import numpy as np
#Read in TU.csv
TU = pd.read_csv("TU.csv")
# pd.set_option('display.max_columns', None)
```

```python
# Divide the dataframe into training and test datasets
# In this course, you will only work on the variables in the training set
# You will need to clean the test set following the methods used in this script, when you work on modeling later.

TUtrain=TU[TU['train-test']=='train']
TUtest=TU[TU['train-test']=='test']
```

```python
#Column1 - ID

#Check NAs
print(TUtrain['ID'].isna().sum())
#No NA,so no cleaning is required.
#add clean to the column name 
TUtrain['ID clean'] = TUtrain['ID']
```

```python
#Column2 - Entry Term Application

#Check NAs
print(TUtrain['Entry Term (Application)'].isna().sum())
#No NA.
TUtrain['Entry Term (Application)'].unique()
#No irregular categories.
#Change column name to include clean
TUtrain['Entry Term (Application) clean'] = TUtrain['Entry Term (Application)']
```

```python
#Column 3 - Admit Type

#Check NAs
print(TUtrain['Admit Type'].isna().sum())
#No NA.
print(TUtrain['Admit Type'].unique())
#No irregular categories.
```

```python
#Since the data set only has first years (i.e.,only one category), 
# Admit.Type should be removed.
print(TUtrain['Admit Type'])
TUtrain=TUtrain.drop('Admit Type',axis='columns')
#All of the columns in TUtrain have been printed to make sure Admit Type was removed 
TUtrain.columns
```

```python
#Column 4 - Permanent Postal
print(TUtrain['Permanent Postal'].isna().sum())
#105 NAs.
TUtrain['Permanent Postal'].unique()
#However, the column "Permanent.Geomarket" had already provided needed information
#regarding the postal codes of different states. Therefore, this column might be
#redundant, and we might just use "Permanent.Geomarket"
#Therefore, let's remove this column and there is no need to handle the missing values
print(TUtrain['Permanent Postal'])
TUtrain=TUtrain.drop('Permanent Postal',axis='columns')
#All of the TUtrain columns have been printed to make sure Permanent Postal has been removed
TUtrain.columns
```

```python
#Column 5 - Permanent Country
#0 NA.
print(TUtrain['Permanent Country'].isna().sum())
#No irregular categories.
TUtrain['Permanent Country'].unique()
#change column name to include clean
TUtrain['Permanent Country clean'] = TUtrain['Permanent Country']
```

```python
#Column6 - Sex
print(TUtrain['Sex'].isna().sum())
#No NA.
print(TUtrain['Sex'].unique())
#No irregular categories.
#change column name to include clean
TUtrain['Sex clean'] = TUtrain['Sex']
```

```python
#Column7 - Ethnicity
print(TUtrain['Ethnicity'].isna().sum())
#158 NAs.
print(TUtrain['Ethnicity'].unique())
#No irregular categories.
# It is fair to replace NAs with "Not Specified" as we do not have other columns for inter-field checking.

TUtrain['Ethnicity']=TUtrain['Ethnicity'].fillna("Not specified")
TUtrain['Ethnicity'].isna().sum()
#There are no more nas after filling them with Not specified 
#renamin the column with clean 
TUtrain['Ethnicity clean'] = TUtrain ['Ethnicity']
```

```python
#Column8 - Race
print(TUtrain['Race'].isna().sum())
#389 NAs.
print(TUtrain['Race'].unique())
```

```python
#No irregular categories.
#Impute NAs with "Not specified", similar to what we do for Ethnicity.
TUtrain["Race"] = TUtrain['Race'].fillna("Not specified")
```

```python
#The current classification of Race is too detailed, which leads to very 
#low frequencies for some categories.Let's take a look at the value frequencies.
TUtrain['Race'].value_counts()
```

```python
#So let's combine some of the categories because a category with a small number of cases won't have 
#a significant effect on the target variable in the modeling stage.

#Group labels with a low number of cases (< 100 cases) into "Others".
# first create a list for races with counts less than 100
races_counts =TUtrain['Race'].value_counts()
races_less_than_100 =races_counts[races_counts<100].index.tolist()
races_less_than_100
```

```python
#If the race category is found within the list then it is grouped into Others
TUtrain['Race'] = TUtrain['Race'].apply(lambda x: 'Others' if x in races_less_than_100 else x)
#The value counts of Race were printed to check if the grouping was done correctly 
print(TUtrain['Race'].value_counts())
#add clean to the column name
TUtrain['Race clean'] = TUtrain['Race']
```

```python
#Column 9 - Religion
# print # of NAs
print(TUtrain['Religion'].isna().sum())
# print unique values for Religion
print(TUtrain["Religion"].unique())
#Making a copy of the column to clean
TUtrain["Religion clean"] = TUtrain["Religion"]
#Impute NAs with "Not specified", similar to what we do for Race.
TUtrain["Religion clean"] = TUtrain['Religion clean'].fillna("Not specified")
print(TUtrain['Religion clean'].isna().sum())
#There are no more NAs
#The current classification of Race is too detailed, which leads to very 
#low frequencies for some categories.Print the value frequencies.
#The value counts were saved to religion counts and printed
religion_counts =TUtrain['Religion clean'].value_counts()
print(religion_counts)

#Religion has lots of categories, with some categories having a very small number of cases. 
#Let's combine similar levels into one level( for example:['Bible Churches','Christian Reformed','Christian Scientist','Church of Christ','Church of God'] )
#then combine levels with less than 100 cases into "Other" because a level accounting for lower than 1% 
#of training set is very unlikely to have a significant effect on the target variable.
#religions that had less than a 100 values were saved to religions_less_than_100
religions_less_than_100 =religion_counts[religion_counts<100].index.tolist()
print("Religions less than a 100:", religions_less_than_100)
#If x is in religions_less_than_100 then it was grouped into 'Other'
TUtrain['Religion clean'] = TUtrain['Religion clean'].apply(lambda x: 'Other' if x in religions_less_than_100 else x)
#value counts were printed to make sure the grouping worked and it did
TUtrain['Religion clean'].value_counts()
```

```python
#Column 10 - First_Source Origin First Source Date
print(TUtrain['First_Source Origin First Source Date'].isna().sum())
#No NAs.

#convert to date format
TUtrain['First_Source Origin First Source Date']= pd.to_datetime(TUtrain['First_Source Origin First Source Date'],format="mixed")
TUtrain['First_Source Origin First Source Date']
```

```python
#Column11 - Inquiry Date
print(TUtrain['Inquiry Date'].isna().sum())
#3181 NAs.

#convert to date format
TUtrain['Inquiry Date']= pd.to_datetime(TUtrain['Inquiry Date'], errors='coerce')
```

```python
#Column 12 - Submitted
print(TUtrain['Submitted'].isna().sum())
#No NAs.

#convert to date format
TUtrain['Submitted']=pd.to_datetime(TUtrain['Submitted'], errors='coerce')
```

```python
#Column10-12
#After viewing Column10-12, it would be interesting to see
#whether the differences between submission date and First_Source date
#the differences between submission date and inquiry date affect the response variable.
#So let's calculate the time difference between submission date and first_source date.
#TUtrain['Submit_FirstSource'] = (TUtrain['Submitted']-TUtrain['First_Source Origin First Source Date']).astype('timedelta64[W]')
#TUtrain['Submit_Inquiry']= (TUtrain['Submitted']-TUtrain['Inquiry Date']).astype('timedelta64[W]')
TUtrain['Submit_FirstSource'] = (TUtrain['Submitted'] - TUtrain['First_Source Origin First Source Date']).dt.days / 7
TUtrain['Submit_Inquiry'] = (TUtrain['Submitted'] - TUtrain['Inquiry Date']).dt.days / 7
```

```python
#There are NAs in Inquiry.Date, 
#thus leading to NAs in Submit_Inquiry.
#Impute NAs in Submit_Inquiry with mean.

TUtrain ['Submit_Inquiry'] = TUtrain['Submit_Inquiry'].fillna(TUtrain['Submit_Inquiry'].mean())
TUtrain['Submit_Inquiry'].isna().sum()
```

```python
#Remove Column10-12 after you created new variables above.   

TUtrain.drop('Submitted',axis='columns', inplace=True)
TUtrain.drop('Inquiry Date',axis='columns', inplace=True)
TUtrain.drop('First_Source Origin First Source Date',axis='columns', inplace=True)
#The columns were printed to make sure these columns were actuall dropped
TUtrain.columns
#add clean to the new column names that were created
TUtrain['Submit_FirstSource clean'] = TUtrain['Submit_FirstSource']
TUtrain['Submit_Inquiry clean'] = TUtrain['Submit_Inquiry']
```

```python
#Column16 - Legacy
print(TUtrain['Legacy'].isna().sum())
#No NAs.
print(TUtrain['Legacy'].unique())
#No irregular categories.
#Impute NAs with "No Legacy"
TUtrain ['Legacy'] = TUtrain['Legacy'].fillna("No Legacy")
TUtrain['Legacy'].isna().sum()
#There are no more null values 
#Legacy has many options, leading some options to having only a small number of cases.
#Let's group all the options into 3 categories ('Legacy',"No Legacy", "Legacy, Opt Out") 
#so that each category has the chance to affect the response variable.
TUtrain['Legacy']=TUtrain['Legacy'].apply(lambda x: 'Legacy, Opt Out' if x not in ['Legacy','No Legacy'] else x)
#Value counts were printed to make sure the grouping was done correctly and it did
TUtrain['Legacy'].value_counts()
#Add clean to the column name
TUtrain['Legacy clean'] = TUtrain['Legacy']
TUtrain['Legacy clean'].value_counts()
```

```python
#Column17 - Athlete
# print # NAs.
print(TUtrain['Athlete'].isna().sum())
#There are 8683 NAs
# print unique value counts.
print(TUtrain['Athlete'].unique())
print(TUtrain['Athlete'].value_counts())
#making a copy of the column to clean
TUtrain ['Athlete clean'] = TUtrain['Athlete']
#Impute NAs with "Non-Athlete"
TUtrain ['Athlete clean'] = TUtrain['Athlete clean'].fillna("Non-Athlete")
TUtrain['Athlete clean'].isna().sum()
#There are no more null values 
#Similar to Legacy, Athlete has many categories with a few cases.
#Group all options into three categories: 
#Athlete, Non-Athlete, and Athlete, Opt Out.
# I made a list of all the categories that should be in Athlete, Opt Out and said that if x is in those they should be grouped into Athlete, Optout and if x isnt in that or Non-Athlete  they should be grouped into Athlete 
opt_out = ['Athlete, Opt Out', 'Athlete, Legacy, Opt Out', ' Athlete, Legacy, Opt Out, VIP', 'Athlete, Fine Arts, Opt Out', 'Athlete, Fine Arts, Legacy, Opt Out, VIP']
TUtrain['Athlete clean']=TUtrain['Athlete clean'].apply(lambda x: 'Athlete, Opt Out' if x in opt_out else x)
TUtrain['Athlete clean']=TUtrain['Athlete clean'].apply(lambda x: 'Athlete' if x not in ['Athlete, Opt Out','Non-Athlete'] else x)
#Value counts were printed to make sure the grouping was done correctly and it was 
print(TUtrain['Athlete clean'].value_counts())
```

```python
#Column18 - Sport 1 Sport

# print # NAs.
print(TUtrain['Sport 1 Sport'].isna().sum())
#There are 8683 NAs
# print unique value counts.
print(TUtrain['Sport 1 Sport'].unique())
print(TUtrain['Sport 1 Sport'].value_counts())
#Making a copy of the column to clean 
TUtrain ['Sport 1 Sport clean'] = TUtrain['Sport 1 Sport']
#Impute NAs with "No Sport"
TUtrain ['Sport 1 Sport clean'] = TUtrain['Sport 1 Sport clean'].fillna("No Sport")
TUtrain['Sport 1 Sport clean'].isna().sum()
#There are no more null values 
#Group sport men and sport women into one group
#so that each group has sufficient cases to have an impact on the response.
sports = TUtrain['Sport 1 Sport clean'].unique()
# Empty lists for mens_sports and womens_sports were made 
mens_sports = [ ]
womens_sports = [ ]
# if the substring 'Men' was found, it should be appended to mens_sports and if the substring 'Women' was found it should be appended to womens_sports and non was found it hsould stay as is 
for text in sports:
    if 'Men' in text:
        mens_sports.append(text)
    elif 'Women' in text:
        womens_sports.append(text)
    else:
        sports
print(mens_sports)
print(womens_sports)
#If a category was in womens_sports it should be grouped into the catgeory 'Sport Women' and if category was in men's sports it should be grouped into the category 'Sport Men'
TUtrain['Sport 1 Sport clean']=TUtrain['Sport 1 Sport clean'].apply(lambda x: 'Sport Women' if x in womens_sports else x)
TUtrain['Sport 1 Sport clean']=TUtrain['Sport 1 Sport clean'].apply(lambda x: 'Sport Men' if x in mens_sports else x)
#Value counts were printed to show the grouping was done correctly and it was
print(TUtrain['Sport 1 Sport clean'].value_counts())
```

```python
#Column24 - Academic Interest 1

#check NAs
print(TUtrain['Academic Interest 1'].isna().sum())
# 4 NAs.
print(TUtrain['Academic Interest 1'].unique())

```

```python
# Step1: Most of the NAs for Academic.Interest.1 have a value for Academic.Interest.2
#We may assign the corresponding values in Academic.Interest.2 
#to NAs in Academic.Interest.1 if Academic.Interest.2 has a value.

# When update values in a subset of dataframes, 
# Try using .loc[row_indexer,col_indexer] = value instead to avoid chained indexing issue
#Make a copy of the column to clean
TUtrain['Academic Interest 1 clean'] = TUtrain['Academic Interest 1']
for index,row in TUtrain.iterrows():
    if pd.isna(row['Academic Interest 1 clean']):
        #print(index,row['Academic Interest 1'],row['Academic Interest 2'])
        TUtrain.loc[index,'Academic Interest 1 clean']=TUtrain.loc[index,'Academic Interest 2']
```

```python
    
# Step2:For the remaining NAs in Academic.Interest.1, assign Undecided.
TUtrain['Academic Interest 1 clean'] = TUtrain['Academic Interest 1 clean'].fillna('Undecided')
TUtrain['Academic Interest 1 clean'].unique()
```

```python

#  Step3: Group Business related options into "Business".
# find all business related labels and group them into business
#A list of the business related options was made
business_related_options = ['Finance', 'Business - Communication Management', 'Economics', 'Business - Accounting', 'Entrepreneurship', 'Business - Sport Management', 'Business - Management','Communication',
                            'Business - Marketing', 'Business - International Business', 'Business Analytics & Technology', 'Business - Management Information Systems', 'Business Legal Studies']

#If the category was in the list it was grouped into 'Business'
TUtrain['Academic Interest 1 clean'] = TUtrain['Academic Interest 1 clean'].apply(lambda x: 'Business' if x in business_related_options else x)
#Value counts were printed to make sure the grouping worked and it did 
#value counts were saved to academic_interest_1_counts
academic_interest_1_counts = TUtrain['Academic Interest 1 clean'].value_counts()
print(academic_interest_1_counts)
# step 4: Group majors with a low number of cases (< 100 cases) into "Others".
# first create a list for majors with counts less than 100
#If an academic interest 1 has less than 100 value counts it was put into interests_less_than_100
interests_less_than_100 =academic_interest_1_counts[academic_interest_1_counts<100].index.tolist()
print("Acadaemic Interests less than a 100:", interests_less_than_100)
#If x was in the interest_less_than_100 list it was grouped into the 'Other' catgeory
TUtrain['Academic Interest 1 clean'] = TUtrain['Academic Interest 1 clean'].apply(lambda x: 'Other' if x in interests_less_than_100 else x)
# value counts were prinetd to ensure it worked and it did
TUtrain['Academic Interest 1 clean'].value_counts()


```

```python
#Column25 - Academic.Interest.2

#Check NAs
print(TUtrain['Academic Interest 2'].isna().sum())
#There are 94 NAs
print(TUtrain['Academic Interest 2'].value_counts())
print(TUtrain['Academic Interest 2'].unique())
#Make a copy of the column to clean
TUtrain['Academic Interest 2 clean'] = TUtrain['Academic Interest 2']
#Replace repeated academic interest 2 with Undecided, which means academic interest 2==academic interest 1

for index,row in TUtrain.iterrows():
    if TUtrain.loc[index,'Academic Interest 2 clean']== TUtrain.loc[index,'Academic Interest 1']:
        TUtrain.loc[index,'Academic Interest 2 clean'] = 'Undecided'
        
print(TUtrain['Academic Interest 2 clean'].value_counts())
#then replace NAs Undecided

TUtrain['Academic Interest 2 clean'] = TUtrain['Academic Interest 2 clean'].fillna('Undecided')


#Group Business related options into "Business".
#A list for the business rleated options was created
business_related_options = ['Finance', 'Business - Communication Management', 'Economics', 'Business - Accounting', 'Entrepreneurship', 'Business - Sport Management', 'Business - Management','Communication',
                            'Business - Marketing', 'Business - International Business', 'Business Analytics & Technology', 'Business - Management Information Systems', 'Business Legal Studies']
#if an option was in that list it was grouped into 'Business'
TUtrain['Academic Interest 2 clean'] = TUtrain['Academic Interest 2 clean'].apply(lambda x: 'Business' if x in business_related_options else x)
# value counts for academic interests 2 was saved to academic_interests_2_counts
academic_interest_2_counts = TUtrain['Academic Interest 2 clean'].value_counts()
print(academic_interest_2_counts)

#Group options with a low number of cases (< 100 cases) into "Others".
# A list with options that had less than a 100 values was made 
#If any of the 2nd academic interests had a count of less than 100 then it was put into interests2_less_than_100
interests2_less_than_100 =academic_interest_2_counts[academic_interest_2_counts<100].index.tolist()
print("Acadaemic Interests less than a 100:", interests2_less_than_100)
# If the category was in the list it was grouped into 'Other'
#Anything in interests2_less_than_100 was grouped into 'Other'
TUtrain['Academic Interest 2 clean'] = TUtrain['Academic Interest 2 clean'].apply(lambda x: 'Other' if x in interests2_less_than_100 else x)
#I printed value counts to ensure the grouping worked and it did 
TUtrain['Academic Interest 2 clean'].value_counts()

```
