## SalePrice Cleaning 
The purpose of this is to find outliers using the 1.5 x IQR rule (boxplot) rule. The output shows us the the IQR bounds (4.79 and 5.66) and that there are 28 outliers. It also show us the first few outliers within the column. I used the 1.5 x IQR rule because I planned on winsorizing to clean the outliers. 

```python
Q1 = Train["SalePrice"].quantile(0.25)
Q3 = Train["SalePrice"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = Train[(Train["SalePrice"] < lower) | (Train["SalePrice"] > upper)]

print("IQR bounds:", lower, "to", upper)
print("Number of outliers:", len(outliers))
outliers[["SalePrice"]].head()
```

## Sale Price Cleaning 
The purpose of this code is to windsorize SalePrice, which caps the outliers to the upper and lower bounds. I chose to windsorize because there are a lot of outliers around the 1st and 3rd quartiles. The output shows us the values of the upper and lower bounds (4.79 and 5.66)  as well as the mean (5.22). 

```python
Train["SalePrice"] = Train["SalePrice"].clip(lower, upper)

print("\nWINSORIZING results:")
print("Outliers capped to the cutoff values of ", str(lower), "and ", str(upper))
print(Train.SalePrice.mean().round(2))
```

## SalePrice Cleaning
The purpose of this code is to generate a box plot of SalePrice after cleaning to make sure there are no more outliers. The output of this code shows that there are no more outliers.

## SalePrice Cleaning

```python
Train["SalePrice"] = zscore(Train.SalePrice)
print('Standardized SalePrice', Train.SalePrice.mean().round())
```

## LotFrontage Cleaning
The purpose of this code is to use linear regression to predict the NAs. It predicts the null values in LotFrontage based off the association between LotFrontage and BldgType_2fmCon, BldgType_Duplex, BldgType_Twnhs, and BldgType_TwnhsE. I chose to use these 4 building types because I talked to Dr. Young and she said LotFrontage and BldgType are closely related. She said not to include BldgType_1Fma because it makes all the BldgTypes perfectly correlated and we don't want that. The output of the code shows us the regression results. 

```python
import pingouin as pg
lotfrontage_data = Train[Train["LotFrontage"].notnull()]
lotfrontage_missing = Train[Train["LotFrontage"].isnull()]
regression_results = pg.linear_regression(
    X = lotfrontage_data[["BldgType_2fmCon","BldgType_Duplex","BldgType_Twnhs","BldgType_TwnhsE"]],
    y = lotfrontage_data["LotFrontage"]
)
print("REGRESSION RESULTS:")
print(regression_results)
```

## LoFrontage Cleaning
The purpose of this code is to use linear regression to predict the null values. It predicts the null values in LotFrontage based off the association between LotFrontage and BldgType_2fmCon, BldgType_Duplex, BldgType_Twnhs, and BldgType_TwnhsE. The output of the code shows that the linear regression works because there are 0 null values after using it. 

```python
intercept = regression_results.loc[0, 'coef'] 
slope1 = regression_results.loc[1, 'coef'] 
slope2 = regression_results.loc[2, 'coef']
slope3 = regression_results.loc[3, 'coef']
slope4 = regression_results.loc[4, 'coef']
predicted_values = intercept + slope1 * lotfrontage_missing['BldgType_2fmCon']  + slope2 * lotfrontage_missing['BldgType_Duplex'] + slope3 * lotfrontage_missing['BldgType_Twnhs'] + slope4 * lotfrontage_missing['BldgType_TwnhsE']
Train.loc[Train['LotFrontage'].isnull(), 'LotFrontage'] = predicted_values
print("Null counts:\n", Train.LotFrontage.isna().sum())
```

## GarageArea Cleaning
The purpose of this is to find outliers in GarageArea using the 1.5 x IQR rule (boxplot) rule. The output shows us the the IQR bounds (4.997 and 7.17) and that there are 84 outliers. It also show us the first few outliers within the column. I used the 1.5 x IQR rule because I planned on winsorizing to clean the outliers.

```python
Q1 = Train["GarageArea"].quantile(0.25)
Q3 = Train["GarageArea"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = Train[(Train["GarageArea"] < lower) | (Train["GarageArea"] > upper)]

print("IQR bounds:", lower, "to", upper)
print("Number of outliers:", len(outliers))
outliers[["GarageArea"]].head()
```

## GarageArea Cleaning 
The purpose of this code is to windsorize GarageArea, which caps the outliers to the upper and lower bounds. I chose to windsorize because there are a lot of outliers around the 1st and 3rd quartiles. The output shows us the values of the upper and lower bounds as well as the mean.

```python
Train["GarageArea"] = Train["GarageArea"].clip(lower, upper)

print("\nWINSORIZING results:")
print("Outliers capped to the cutoff values of ", str(lower), "and ", str(upper))
print(Train.GarageArea.mean().round(2))
```

I am finding the two errors around -2.5 and setting the GrLivArea for those two errors to 0, which is the mean of the GrLivArea. This will allow me to pass the third assumption.

```python
predictors = Train[["TotalBsmtSF","Fireplaces_0","Fireplaces_2", "KitchenQual_Ex","KitchenQual_TA","GarageCars_1", "GarageCars_3","GrLivArea", "BldgType_Twnhs",
                    "BldgType_TwnhsE", "OverallQual_7", "OverallQual_8", "OverallQual_9", "OverallQual_10", "LotArea","GarageType_NoGarage", 
                    "GarageType_Detchd","Neighborhood_NridgHt", "Neighborhood_NoRidge","Neighborhood_StoneBr", "FullBath_3", "FullBath_2",
                    "Neighborhood_Edwards","Neighborhood_BrDale", "MSSubClass_60", "MSSubClass_120", "MSSubClass_30","MSSubClass_50","MSSubClass_90","ExterQual_Fa", "YearBuilt"]]
```

```python
model = pg.linear_regression(y = Train.SalePrice, X= predictors) 
print(model)
```

```python
residuals = pd.Series(model.residuals_)
Train["error"] = residuals
Train["calculated_SalePrice"] = Train.SalePrice - residuals
```

The third assumption is passed since the plot shows randomness and there are no more errors close -2.5 since I cleaned up GrLivArea some more. This means that the variance of the residuals is not correlated with the calculated SalePrice values, which shows homoskedasticity. 

```python
vif = pd.DataFrame()
vif['VIF'] = [variance_inflation_factor(predictors.values, i) for i in range(predictors.shape[1])]
vif['variable'] = predictors.columns
vif
```

My VIF results show that none of my independent variables that have high correlations with each other, as they all have a VIF well under 10.
